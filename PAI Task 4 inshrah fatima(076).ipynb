## Dynamic N-Queens

```python
def is_safe(board, row, col, n):
    
    # Check same column
    for i in range(row):
        if board[i][col] == 1:
            return False

    # Check left diagonal
    i = row - 1
    j = col - 1
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Check right diagonal
    i = row - 1
    j = col + 1
    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    return True


def solve_nqueens(board, row, n):

    if row == n:
        print("\nSolution:")
        for i in range(n):
            for j in range(n):
                print("Q" if board[i][j] == 1 else ".", end=" ")
            print()
        return True

    for col in range(n):
        if is_safe(board, row, col, n):

            board[row][col] = 1

            if solve_nqueens(board, row + 1, n):
                return True

            board[row][col] = 0   # Backtrack

    return False


# Dynamic input
n = int(input("Enter number of Queens: "))

board = [[0 for _ in range(n)] for _ in range(n)]

if not solve_nqueens(board, 0, n):
    print("No solution exists")
```
