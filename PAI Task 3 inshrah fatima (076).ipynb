```python
# Water Jug Problem using DFS

jug1_capacity = 4
jug2_capacity = 3
goal = 2

visited = set()

def dfs(a, b, path):
    
    if (a, b) in visited:
        return False

    visited.add((a, b))

    path.append((a, b))

    # Goal condition
    if a == goal:
        print("Goal Reached!\n")
        print("Steps:")
        for step in path:
            print(step)
        return True

    # Rule 1: Fill Jug A
    if dfs(jug1_capacity, b, path.copy()):
        print("Rule Applied: Fill Jug A")
        return True

    # Rule 2: Fill Jug B
    if dfs(a, jug2_capacity, path.copy()):
        print("Rule Applied: Fill Jug B")
        return True

    # Rule 3: Empty Jug A
    if dfs(0, b, path.copy()):
        print("Rule Applied: Empty Jug A")
        return True

    # Rule 4: Empty Jug B
    if dfs(a, 0, path.copy()):
        print("Rule Applied: Empty Jug B")
        return True

    # Rule 5: Pour A → B
    transfer = min(a, jug2_capacity - b)
    if dfs(a - transfer, b + transfer, path.copy()):
        print("Rule Applied: Pour A → B")
        return True

    # Rule 6: Pour B → A
    transfer = min(b, jug1_capacity - a)
    if dfs(a + transfer, b - transfer, path.copy()):
        print("Rule Applied: Pour B → A")
        return True

    return False


# Start from (0,0)
dfs(0, 0, [])
```
